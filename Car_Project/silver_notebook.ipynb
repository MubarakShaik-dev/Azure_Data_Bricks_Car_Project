```python
from pyspark.sql.functions import *
```

```python
df = spark.read.format('parquet')\
    .option('inferSchema', 'true')\
    .load('abfss://bronze@adlscarproject.dfs.core.windows.net/rawData')
```

```python
df =df.withColumn('Model_Category',split(col('Model_ID'),'-')[0])
```

```python
df.withColumn('Units_Sold',col('Units_Sold').cast('string'))
```

```python
df = df.withColumn('Revenue_PerUnit',col("Revenue")/col("Units_Sold"))
```

## Data Writing

```python
df.write.format('parquet')\
    .mode('overwrite')\
        .option('path','abfss://silver@adlscarproject.dfs.core.windows.net/carSales')\
            .save()
```
